# 7. Data Preprocessing

The dataset contains information in different formats. Some columns, such as genres and keywords, store their values as JSON strings instead of regular lists.

These fields need to be processed before they can be used for searching and comparing movies.

The preprocessing stage also includes handling missing values in the columns required for the analysis.

### 7.1 Selecting Relevant Columns

Not all 20 columns are necessary for this project. The analysis focuses on movie titles, genres, keywords, descriptions, popularity and ratings.

A smaller selection of columns makes the data easier to work with.

```python
selected_columns = [
    'title',
    'genres',
    'keywords',
    'overview',
    'popularity',
    'vote_average',
    'vote_count',
    'release_date'
]

movies = movies[selected_columns].copy()

movies.head()
```

### 7.2 Handling Missing Values

Some movies do not have an overview or a release date. Missing overviews are replaced with empty strings because this column will be used later for text analysis.

Movies without a release date are kept because this information is not required for the recommendation system.

```python
movies['overview'] = movies['overview'].fillna('')
movies['release_date'] = movies['release_date'].fillna('')
```

### 7.3 Extracting Genres and Keywords

The genres and keywords columns contain structured information stored as JSON strings.

A function is created to extract the names from these fields and convert them into Python lists.

```python
def extract_names(data):
    try:
        items = json.loads(data)
        return [item['name'] for item in items]
    except (TypeError, ValueError):
        return []

movies['genre_list'] = movies['genres'].apply(extract_names)
movies['keyword_list'] = movies['keywords'].apply(extract_names)

movies[['title', 'genre_list', 'keyword_list']].head()
```

After preprocessing, genres and keywords are available as Python lists instead of JSON strings.

This format makes it easier to search for movies by genre and prepare the text data for the similarity analysis.

# 8. Preparing Movie Features

The next step is to combine the most relevant text information into a single column called `tags`.

This column will contain the movie genres, keywords and overview. It will be used later to compare movies and calculate their similarity.

### 8.1 Combining Text Features

The genres and keywords are stored as lists, so they need to be converted into strings before being combined with the movie overview.

```python
movies['tags'] = (
    movies['genre_list'].apply(lambda x: ' '.join(x)) + ' ' +
    movies['keyword_list'].apply(lambda x: ' '.join(x)) + ' ' +
    movies['overview']
)

movies[['title', 'tags']].head()
```

```python
movie_records = movies.to_dict('records')
```

The new `tags` column contains the main textual information about each movie.

For example, movies with similar genres, keywords or descriptions may have similar tags. This information will be useful for the similarity analysis.

# 9. Searching in Movie Data

Searching is the process of finding specific information in a collection of data.

In this project, searching is used to find movies based on their titles or genres. Python functions and filtering techniques make it possible to check which movies match a given condition.

### 9.1 Searching Movies by Title

The first search function finds movies based on their titles. It also allows partial searches, so the complete movie title is not required.

The search is case-insensitive, meaning that uppercase and lowercase letters do not affect the results.

```python
def search_by_title(title):
    results = list(filter(
        lambda movie: title.lower() in movie['title'].lower(),
        movie_records
    ))

    columns = ['title', 'genre_list', 'vote_average', 'popularity']

    if not results:
        return pd.DataFrame(columns=columns)

    return pd.DataFrame(results)[columns]

search_by_title('batman')
```

The `filter()` function checks each movie and keeps only the records that match the condition.

The lambda expression defines the condition without creating a separate function. This makes the search shorter and easier to reuse.

### 9.2 Searching Movies by Genre

The next function searches for movies that belong to a selected genre.

Since each movie can have multiple genres, the function checks whether the selected genre exists in the movie's genre list.

```python
def search_by_genre(genre):
    results = list(filter(
        lambda movie: genre.lower() in [g.lower() for g in movie['genre_list']],
        movie_records
    ))

    columns = ['title', 'genre_list', 'vote_average', 'popularity']

    if not results:
        return pd.DataFrame(columns=columns)

    return pd.DataFrame(results)[columns].head(10)

search_by_genre('comedy')
```

### 9.3 Testing Searching Functions

To check if the searching functions work correctly, several examples are tested using different search terms.

The tests include a movie title that exists, a genre with multiple matching movies and a title that does not exist in the dataset.

```python
# Test 1: Search by title
title_results = search_by_title('batman')

print("TEST 1 - SEARCH BY TITLE")
print(f"Number of results: {len(title_results)}")
display(title_results[['title', 'genre_list']].head(5))


# Test 2: Search by genre
genre_results = search_by_genre('comedy')

print("\nTEST 2 - SEARCH BY GENRE")
print(f"Number of results: {len(genre_results)}")
display(genre_results[['title', 'genre_list']].head(5))


# Test 3: Search for a non-existing movie
missing_results = search_by_title('nonexistentmovie')

print("\nTEST 3 - UNSUCCESSFUL SEARCH")

if missing_results.empty:
    print("No movies found for the selected title.")
```

### 9.4 Results and Observations

The tests show that both searching functions return the expected results.

The title search successfully finds movies containing the word "batman", including partial matches. The genre search returns movies that belong to the Comedy category, even when they have other genres as well.

The unsuccessful search also works correctly. When no matching title is found, the function returns an empty result instead of producing an error.

# 10. Sorting in Movie Data

Sorting is the process of arranging data in a specific order. In a movie dataset, sorting can help users find the most popular or highest-rated films.

Python provides the `sorted()` function, which can organize collections based on a selected attribute. The `key` parameter defines the value used for sorting, while `reverse=True` changes the order from ascending to descending.

### 10.1 Sorting Movies by Popularity

Popularity is one of the attributes available in the dataset. Sorting movies by this value makes it possible to identify the most popular titles.

```python
popular_movies = sorted(
    movie_records,
    key=lambda movie: movie['popularity'],
    reverse=True
)

pd.DataFrame(popular_movies)[
    ['title', 'popularity', 'vote_average']
].head(10)
```

The results show the ten movies with the highest popularity values.

The `lambda` expression tells `sorted()` to use the popularity attribute when comparing movies. The `reverse=True` argument places the highest values first, in descending order.

### 10.2 Sorting Movies by Rating

Movies can also be ranked based on their average user ratings.

To make the results more reliable, movies with fewer than 100 votes are excluded. This avoids placing films with very few ratings at the top of the list.

```python
rated_movies = list(filter(
    lambda movie: movie['vote_count'] >= 100,
    movie_records
))

top_rated_movies = sorted(
    rated_movies,
    key=lambda movie: movie['vote_average'],
    reverse=True
)

pd.DataFrame(top_rated_movies)[
    ['title', 'vote_average', 'vote_count']
].head(10)
```

This ranking shows the highest-rated movies among titles with at least 100 user votes.

The filtering step helps reduce the influence of movies that have received very few ratings.

### 10.3 Testing Sorting Functions

The following tests check whether both sorting methods correctly arrange movies in descending order.

```python
# Test 1: Check popularity order
print("TEST 1 - SORTING BY POPULARITY")

print("Correct order:",
      popular_movies == sorted(
          popular_movies,
          key=lambda movie: movie['popularity'],
          reverse=True
      ))

display(pd.DataFrame(popular_movies)[['title', 'popularity']].head(5))
```

```python
# Test 2: Check rating order
print("TEST 2 - SORTING BY RATING")

print("Correct order:",
      top_rated_movies == sorted(
          top_rated_movies,
          key=lambda movie: movie['vote_average'],
          reverse=True
      ))

display(pd.DataFrame(top_rated_movies)[
    ['title', 'vote_average', 'vote_count']
].head(5))
```

### 10.4 Results and Observations

The tests confirmed that both sorting methods work correctly.

The popularity sorting identified movies such as Minions, Interstellar and Deadpool as some of the most popular titles in the dataset.

The rating sorting placed highly rated movies such as The Shawshank Redemption, The Godfather and Fight Club at the top. By including only movies with at least 100 votes, the results are less affected by ratings based on very few users.

These examples show how sorting can help organize movie data and identify relevant results based on different criteria.

# 11. Similarity in Movie Data

Similarity measures how close two or more objects are based on their characteristics.

In this project, similarity is used to find movies with similar content. The comparison is based on genres, keywords and movie descriptions.

This approach allows the recommendation system to suggest films that share important characteristics with a selected movie.

### 11.1 Text Vectorization

Computers cannot directly compare text in the same way people understand it. Therefore, movie descriptions need to be converted into numerical data.

For this purpose, the CountVectorizer method from Scikit-learn is used. It converts text into a matrix where each column represents a word and each row represents a movie.

The values show how many times each word appears in the movie's text.

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')

vectors = cv.fit_transform(movies['tags']).toarray()

print("Vector matrix shape:", vectors.shape)
```

### 11.2 Cosine Similarity

After converting the movie text into numerical vectors, the next step is to compare them.

Cosine Similarity measures how similar two vectors are by calculating the cosine of the angle between them.

A value closer to 1 indicates greater similarity, while a value closer to 0 indicates less similarity.

This method is used to calculate similarity scores between all movies in the dataset.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print("Similarity matrix shape:", similarity.shape)
```

### 11.3 Building the Recommendation Function

The recommendation function takes a movie title as input and searches for its position in the dataset.

It then retrieves the similarity scores for that movie, sorts the scores in descending order and returns the five most similar movies.

If the requested title does not exist, the function displays a message instead of returning recommendations.

```python
def recommend(movie_title):
    movie_title = movie_title.lower()

    matches = movies[movies['title'].str.lower() == movie_title]

    if matches.empty:
        print("Movie not found!")
        return

    movie_index = matches.index[0]

    movie_scores = list(enumerate(similarity[movie_index]))

    sorted_scores = sorted(
        movie_scores,
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for index, score in sorted_scores[1:6]:
        recommendations.append(movies.iloc[index]['title'])

    return recommendations
```

### 11.4 Testing the Recommendation System

The recommendation function is tested with different movie titles to observe how the system identifies similar films.

The examples include movies from different genres, allowing the results to be compared.

```python
recommend('Avatar')
```

```python
recommend('The Dark Knight')
```

```python
recommend('Titanic')
```

```python
recommend('Unknown Movie')
```

### 11.5 Results and Observations
The recommendation system identified movies with similar content based on their genres, keywords and descriptions.

The recommendations for The Dark Knight were particularly relevant, as most results were other Batman movies. Avatar also returned several science fiction films.

However, the results were not always accurate, showing that text-based similarity does not necessarily capture the actual meaning or story of a movie.

# 13. Limitations

Although the recommendation system works, it has some limitations:

- **Word-based comparison:** CountVectorizer compares words, but does not understand their meaning or context.
- **Limited information:** Recommendations depend on the genres, keywords and overviews available in the dataset.
- **Different types of similarity:** Movies can share keywords without having similar stories.
- **No user preferences:** The system does not consider personal ratings or viewing history.

For example, the recommendations for Titanic show that sharing some words or general themes does not always mean that two movies are truly similar.

A possible improvement would be using more advanced text analysis methods that can understand the meaning of descriptions instead of only comparing individual words.
